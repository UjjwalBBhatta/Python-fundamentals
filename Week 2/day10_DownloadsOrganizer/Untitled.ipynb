```python
#scanner.py
def scan_folder(folder_path):
    """A function to scan through a folder and find out the files"""
    path = Path(folder_path)   #this will get the true path 
    files = [f for f in path.iterdir() if f.is_file()]   #Creates a list of files, used list comprehension
    """"So whats happening is in the main folder it is iterating through all the files and checking wether it is a file
    and if it is one then storing it in files list"""
    return files
```
