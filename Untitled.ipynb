```python
#A simple file analysing CLI

import timeit     # todays exercise also contained benchmark checks
 
# Generator version    (Todays learning objective)
"""We use generators to iterate line by line instead of loading the whole file, think of minecrafts chunks
being loaded, if we try to load all at once its gonna crash"""
def read_file_gen(filepath):
    with open(filepath, "r", encoding="utf-8") as f:  #encodind= "utf-8" ensures that values are read properly
        for line in f:
            yield line.strip()

# Regular version, without using a generator
def read_file_list(filepath):
    with open(filepath, "r", encoding="utf-8") as f:
        return [line.strip() for line in f]  #using simple comprehensions instead of huge loops

# Word counter using comprehensions
def analyze_text(lines):
    words = [word for line in lines for word in line.split()]
    unique = set(words)
    longest = max(words, key=len) if words else ""
    return {
        "lines": len(lines),
        "words": len(words),
        "unique_words": len(unique),
        "longest_word": longest
    }

if __name__ == "__main__":
    file = "sample.txt"

    # Generator version
    stats_gen = analyze_text(list(read_file_gen(file)))
    print("\nGenerator Stats:", stats_gen)

    # List version
    stats_list = analyze_text(read_file_list(file))
    print("\nList Stats:", stats_list)

    # Benchmark performance
    gen_time = timeit.timeit(lambda: list(read_file_gen(file)), number=100)
    list_time = timeit.timeit(lambda: read_file_list(file), number=100)
    print(f"\n⏱ Generator: {gen_time:.4f}s | List: {list_time:.4f}s")
```
